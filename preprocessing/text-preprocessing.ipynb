```python
# stop-words removal
from nltk.corpus import stopwords

text = "I love learning about natural language processing"

tokens = text.split()
stop_words = set(stopwords.words('english'))

filter_words = [
    word for word in tokens
    if word.lower() not in stop_words
]
print(filter_words)
```

```python
# stemming
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "plays", "studies", "running"]

stemmed_words = [
    stemmer.stem(word)
    for word in words
]

print(stemmed_words)
```

```python
# lemmatization
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
lemmatizer = WordNetLemmatizer()
words = ['playing', 'played', 'studies', 'running']

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in words
]
print(lemmatized_words)
```

```python
words = [
    ('playing', 'v'),
    ('played', 'v'),
    ('studies', 'v'),
    ('running', 'v')
]

result = [
    lemmatizer.lemmatize(word, pos)
    for word, pos in words

]

print(result)
```

```python
# handling numbers 

import re
text = 'I have 25 books and 7 laptops'
text = re.sub(r'\d+', '', text)

print(text)
```

```python
# replacing number with <NUMBER>

text = 'I have 25 books and 7 laptops'
text = re.sub(r'\d+', '<NUMBER>', text)
print(text)
```

```python
# handling url
text = "Visit https://example.com for more information."
text = re.sub(
    r'https?://\S+|www\.\S+',
    '<URL>',
    text
)

print(text)
```

```python
# Handling special characters 

import re
text = "Hello @John! #MachineLearning $100 %"

text = re.sub(r'[^a-zA-Z0-9\s]', '', text)

print(text)
```
