- Technique that groups different inflected forms of a word into a single base or dictionary form,known as a lemma.

```python
import string
```

```python
# tokenization
import nltk
from nltk.tokenize import word_tokenize
```

```python
# lemmatization
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
lemmatizer = WordNetLemmatizer()
words = ['playing', 'played', 'studies', 'running', 'walks']

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in words
]
print(lemmatized_words)
```

```python
words = [
    ('playing', 'v'),
    ('played', 'v'),
    ('studies', 'v'),
    ('running', 'v')
]

result = [
    lemmatizer.lemmatize(word, pos)
    for word, pos in words

]

print(result)
```
